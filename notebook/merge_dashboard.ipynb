```python
import pandas as pd

base = pd.read_csv('../data/employee_attrition_clean.csv')

q1 = pd.read_csv('../data/q1_output.csv')
q2 = pd.read_csv('../data/q2_output.csv')
q3 = pd.read_csv('../data/q3_output.csv')
q4 = pd.read_csv('../data/q4_output.csv')

dashboard = base.merge(q1, on='Employee_ID', how='left') \
                 .merge(q2, on='Employee_ID', how='left') \
                 .merge(q3, on='Employee_ID', how='left') \
                 .merge(q4, on='Employee_ID', how='left')

print('Shape:', dashboard.shape)
print(dashboard[['Pay_Flag', 'Attrition_Risk_Flag', 'High_Potential_Flag', 'Segment_Label']].isnull().sum())

dashboard.to_csv('../powerbi/dashboard.csv', index=False)
print('Saved dashboard.csv:', dashboard.shape)
```
